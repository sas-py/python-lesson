### **if～else文による条件分岐**

```python
age = 18
if age < 18:
    print('まだ選挙権はありません')
else:
    print('投票に行きましょう')
```

### **if～elif～else文による条件分岐**

```python
age = 20
if age < 4:
    print('入場料は無料です')
elif age < 13:
    print('子供料金で入場できます')
else:
    print('大人料金が必要です')
```

### **三項演算子**

```python
a = 1
b = 10
c = a if a > b else b

print(c)
```

```python
age = 18
if age < 18:
    print('まだ選挙権はありません')
else:
    print('投票に行きましょう')
```
