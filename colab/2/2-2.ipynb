### **while文による処理の繰り返し**

```python
i = 0
while i < 5:
    print('こんにちは')
    i += 1
```

### **while文の例**

```python
i = 20
while i > 0:
    print(i)
    i -= 5
```

### **for文による処理の繰り返し**

```python
for i in [10, 20, 30, 40, 50]:
    print(i)
```

### **for文の処理の流れ**

```python
for i in ['spring', 'summer', 'autumn', 'winter']:
    print(i)
```

### **ループ処理の流れの変更**

break：ループの処理を中断する

```python
total = 0
for i in range(10):
    total += i
    if total > 20:
        break
print(i, total)
```

continue：ループ内の処理をスキップする

```python
total = 0
for i in range(100):
    if i % 3 == 0:
        continue
    print(i)
    total += i

print('合計は', total)
```

### **ループ処理のネスト**

```python
for a in range(1, 4):
    print('a=', a)
    for b in range(1, 4):
        print('    b=', b)
```
