<h4 align="left"><strong>(2) 実践演習</strong></h4>
支店別売上データがCSVファイルで格納されている。<br>
このファイルを読み込み売上達成、未達を表示して各支店別売上グラフを表示する。

支店別売上データ：<br>
ファイル名：sales_data.csv<br>

Tokyo,10,20,30<br>
Osaka,40,50,60<br>
Kyoto,70,80,90<br>
<br>

売上達成、未達条件：<br>
売上合計が100を超えていたら目標達成と表示する<br>
超えていなければ目標未達と表示する<br>
<br>
実行結果：<br>
Tokyoは目標未達。 合計: 60<br>
✨ Osakaは目標達成！ 合計: 150<br>
✨ Kyotoは目標達成！ 合計: 240<br>
<br>

(2)実践演習実行結果グラフ.png


```python
import csv
import matplotlib.pyplot as plt

!wget -O sales_data.csv https://raw.githubusercontent.com/sas-py/python-lesson/main/colab/3/sales_data.csv

# 1. データの読み込みと準備
branches = []      # 支店名をいれる空リスト
total_sales = []   # 合計売上をいれる空リスト

with open('sales_data.csv', 'r', encoding='utf-8') as f:
    reader = csv.reader(f)
    for row in reader:
        # row[0]は支店名、row[1:]は売上
        branches.append(row[0])

        # 繰り返しを使って、文字列から整数（int）に変換
        sales_nums = []
        for val in row[1:]:
            sales_nums.append(int(val))

        # 条件分岐：もし売上の合計が100を超えていたらボーナス計算（実務のif練習）
        total = sum(sales_nums)
        if total > 100:
            print(f"✨{row[0]}は目標達成！ 合計: {total}")
        else:
            print(f"   {row[0]}は目標未達。 合計: {total}")

        total_sales.append(total)

# 2. グラフの作成（matplotlibの体験）
plt.bar(branches, total_sales, color='skyblue')
plt.title('Branch Sales Report')
plt.xlabel('Branch')
plt.ylabel('Sales')
plt.show()
```

## **（2）データ読み込みとグラフ表示処理を関数を使用した処理に変更する**

```python
import csv
import matplotlib.pyplot as plt

!wget -O sales_data.csv https://raw.githubusercontent.com/sas-py/python-lesson/main/colab/3/sales_data.csv

def get_total_sales():
    # 1. データの読み込みと準備
    branches = []      # 支店名をいれる空リスト
    total_sales = []   # 合計売上をいれる空リスト

    with open('sales_data.csv', 'r', encoding='utf-8') as f:
        reader = csv.reader(f)
        for row in reader:
            # row[0]は支店名、row[1:]は売上
            branches.append(row[0])

            # 繰り返しを使って、文字列から整数（int）に変換
            sales_nums = []
            for val in row[1:]:
                sales_nums.append(int(val))

            # 条件分岐：もし売上の合計が100を超えていたらボーナス計算（実務のif練習）
            total = sum(sales_nums)
            if total > 100:
                print(f"✨{row[0]}は目標達成！ 合計: {total}")
            else:
                print(f"   {row[0]}は目標未達。 合計: {total}")

            total_sales.append(total)

    return branches, total_sales

def draw_total_sales(branches, total_sales):
    # 2. グラフの作成（matplotlibの体験）
    plt.bar(branches, total_sales, color='skyblue')
    plt.title('Branch Sales Report')
    plt.xlabel('Branch')
    plt.ylabel('Sales')
    plt.show()

# 1. データの読み込みと準備
branches, total_sales = get_total_sales()

# 2. グラフの作成（matplotlibの体験）
draw_total_sales(branches, total_sales)
```
