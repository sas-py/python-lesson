## 関数の定義
**関数**は処理（手続きの流れ）をまとめた再利用可能なコードです。関数には以下の特徴があります。

* 名前を持つ。
* 手続きの流れを含む。
* 返値（明示的あるいは非明示的に）を返す。

`print()`、`id()`、`len()`、`del()`、`sum()` などの[組み込み関数](https://docs.python.jp/3/library/functions.html)は関数の例です。

まず、関数の定義をしてみましょう。関数を定義するには **`def`** を用います。

>**構文：関数の定義**  
>def 関数名():  
>&emsp;&emsp;処理内容  

>関数名()
* このように定義した関数を呼び出すには関数名()と記述します。

### **ユーザー定義関数**
関数の名前は自由につけることができますが、監修としてすべて小文字にします。  
変数名と同様に、複数の単語を含む名前にするときには、**do_something**のように単語を<b>_</b>でつなげます。

```python
def say_hello():
    print('こんにちは')
    print('今日はよい天気ですね')

say_hello()
say_hello()
```

### **関数呼び出しの処理の流れ**


```python
def function_a():
    print('function_aの内部の命令文です')

print('これからfunction_aを呼び出します')
function_a()
print('function_aの呼び出しが終わりました')
```

```python
def function_a():
    print('function_aの処理です')

def function_b():
    print('function_bの処理です')

function_a()
function_b()
```

### **関数の定義位置**  
*   関数の定義は、関数の呼び出しの前に行われている必要があります。


```python
def function_a():
    print('function_aの処理です')

function_a()
function_b()

def function_b():
    print('function_bの処理です')

```

### **関数の呼び出しの階層**  

```python
def function_a():
    print('function_aの処理です')

def function_b():
    print('function_b の処理開始')
    function_a()
    print('function_bの処理終了')

print('function_bを呼び出します')
function_b()
```

### **関数と「変数のスコープ」**
*   変数には参照できる範囲（スコープ）がある
*   ローカル変数：関数の中でのみ使用できる



```python
def function_a():
    a = 10
    print(a)

print(a)
```

*   グローバル変数：プログラムコード全体で使用できる

```python
a = 10

def function_a():
    print(a)

function_a()
```

### **関数と「変数のスコープ」**
*   関数の中でグローバル変数の値を変更したい場合は**global**キーワードを使用する
*   global宣言がある場合

```python
a = 10

def function_a():
    global a
    a = 5

function_a()
print(a)
```

*   global宣言がない場合

```python
a = 10

def function_a():
    a = 5

function_a()
print(a)
```

### **関数の引数**
*   引数のない関数  
countdownという名前の関数を定義しています。  
処理内容は、5から0までの値をカウントダウンしながら出力します。  
このcountdown関数は、カウントダウンの始まりが5に固定されています。


```python
def countdown():
    print('カウントダウンをします')
    counter = 5
    while counter >= 0:
        print(counter)
        counter -= 1

countdown()
```

*   引数のある関数
カウントダウンの始まりを必要に応じて10や3など指定した値に変更する。
*   countdown関数内での変数について  
startとcounterの2つの変数を使用しています。  
この2つの変数はどちらもローカル変数で、参照できるのはcountdown関数の中だけです。

```python
def countdown(start):
    print('関数が受け取った値:', start)
    print('カウントダウンをします')
    counter = start
    while counter >= 0:
        print(counter)
        counter -= 1

countdown(3)
countdown(10)
```

### **引数が複数ある関数**
*   関数を呼び出す側  
関数名の後ろの()の中にカンマ(,)で区切って、必要なだけ値を並べます。
*   関数側  
関数名の後ろの()の中に変数名を、必要なだけカンマ(,)で区切って並べます。



```python
def countdown(start, end):
    print('1つ目の引数で受け取った値:', start)
    print('2つ目の引数で受け取った値:', end)
    print('カウントダウンをします')
    counter = start
    while counter >= end:
        print(counter)
        counter -= 1

countdown(7, 3)
```

### **キーワード引数**
*   デフォルト値のある引数は、そうでない引数よりも引数列の後方に書く必要がある。
*   引数列に「変数名=デフォルト値」を記述しておくと、関数を呼び出すときに、引数を省略できる





```python
def countdown(start, end=0):
    print('1つ目の引数で受け取った値:', start)
    print('2つ目の引数で受け取った値:', end)
    print('カウントダウンをします')
    counter = start
    while counter >= end:
        print(counter)
        counter -= 1

countdown(10)
countdown(10, 0)
countdown(10, 5)
```

### **可変長引数**
*   可変長引数：引数の数に制限がない
*   タプルで複数の値を受け取る

```python
def func_a(*args):
    for a in args:
        print(a)

func_a(1, 2)
func_a(1, 2, 3, 4)
```

*   可変長引数をもつaverage関数で、受け取った値（数値）の平均を出力します

```python
def average(*args):
    total = 0
    for a in args:
        total += a
    print(total / len(args))

average(70, 85, 100, 90)
```

*   Pythonには、リストやタプルの要素の総和を求める組み込み関数であるsum関数があります。  
このsum関数を使うとaverage関数は次のようにシンプルにできます。

```python
def average(*args):
    print(sum(args) / len(args))

average(70, 85, 100, 90)
```

*   辞書で複数のキーと値のペアを受け取る

```python
def func_b(**kwargs):
    for k, v in kwargs.items():
        print(k, v)

func_b(a=1, b=2)
func_b(c=3, d=4, e=5, f=6)
```

```python
def print_data(**kwargs):
    for key, value in kwargs.items():
        print(f'キー:{key}, 値:{value}')

print_data(item='リンゴ', count=1, price=120)
```

### **関数の戻り値**
*   関数から値を戻すには関数内の命令文の末尾に**return**というキーワードに続けて、戻り値を記述します。
*   return 値




*   戻り値のある関数




```python
def circle_area(r):
    return r * r * 3.14

s = circle_area(2.5)
print('半径2.5の円の面積は', s)
```


*   真偽値を返す関数



```python
def is_positive(i):
    if i > 0:
        return True
    else:
        return False

a = -10;
if is_positive(a) == True:
    print('aの値は正です')
else:
    print('aの値は負またはゼロです')
```

*   論理式の値が戻り値になる場合や、関数の戻り値が真偽値の場合は、短く簡潔に書くようにします。




```python
def is_positive(i):
    return i > 0

a = -10;
if is_positive(a):
    print('aの値は正です')
else:
    print('aの値は負またはゼロです')
```


*   複数の値を戻す
*   return文によって。値を呼び出し元に戻すことができます。  
ただし、関数に記述できるreturn文は1つだけです。  
複数の値を戻す場合はタプルを利用し複数の値を戻します。





```python
def get_two_numbers():
    # return (2, 3)
    return 2, 3

a, b = get_two_numbers()
print(a, b)
```
