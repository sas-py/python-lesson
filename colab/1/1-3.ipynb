### **リスト**

```python
a = [10, 20, 30, 40, 50]
print(a[0])
print(a[1])
print(a[4])
```

### **マイナスのインデックス**

```python
a = [10, 20, 30, 40, 50]
print(a[-1])
print(a[-2])
print(a[-5])
```

### **リスト内の値の変更**

```python
a = [10, 20, 30, 40, 50]
print(a)

a[0] = 99
print(a)

a[-1] = 'A'
print(a)
```

### **リストの要素数の確認**

```python
a = [10, 20, 30, 40, 50]
print(len(a))
```

### **listクラスの主なメソッド**

```python
a = [10, 20, 30, 40]

a.append(50)
print(a)

a.insert(2,25)
print(a)

a.remove(25)
print(a)

a.pop()
print(a)

a.clear()
print(a)

a = [10, 20, 30, 40]
print(a.index(30))

a = [10, 20, 30, 40, 10, 50, 10, 60]
print(a.count(10))

a =['hello', 'Hello', 'Hello, Python']
print(a.count('Hello'))

a = [10, 20, 30, 40, 50]
a.reverse()
print(a)

a = ['a', 'b', 'c', 'd', 'e']
a.reverse()
print(a)
```

### **メソッドを使ったリストの操作**

```python
a = [10, 20, 30, 40]

a.append(100)
print(a)

a.pop()
print(a)

a.reverse()
print(a)

print(a.index(20))
```

### **メソッド以外のリストの操作**

```python
a = [10, 20, 30, 40]
del a[1]
print(a)

a = a + [0, 1]
print(a)

print(1 in a)
print(2 in a)

len(a)

sorted(a)

print(sorted(a, reverse=True))
```

### **内包表記**

```python
data = [2**n for n in range(1, 11)]
print(data)

print([str(n) + '月' for n in range(1, 13)])
```

### **if構文を含む内包表記**

```python
data0 = ['apple', 'orange', 'banana', 'avocado']
data1 = [s for s in data0 if s[0] == 'a']
print(data0)
print(data1)
```

### **リストを含むリスト**

```python
data = [[1, 2], [3, 4, 5]]
print(data[0][0])
print(data[0][1])
print(data[1][0])
print(data[1][1])
print(data[1][2])
```

```python
data = [
    [1, 2],
    [3, 4, 5]
]

for row in data:
    print(row)
```

```python
data = [
    [1, 2],
    [3, 4, 5]
]

for row in data:
    for colum in row:
        print(colum)
```

### **タプル**
